```python
import requests
import datetime
from time import sleep
from subprocess import Popen
import json
from pathlib import Path

session = requests.session()
session.verify = False
requests.packages.urllib3.disable_warnings()
```

```python
def get_task_list(taskId):
    url = "https://adc-pc.alibaba-inc.com/task/getList"
    payload = {
        "current": 1,
        # "game_id": "9207",
        # "game_platform": "jiuyou",
        "id": taskId,
        "pageSize": 100,
    }
    response = session.post(url, json=payload)
    task_list = response.json()["result"]["data"]["content"]
    return task_list
```

# 获取商品id

```python
taskIdList = []
goodsIdList = []
for taskId in taskIdList:
    task_info = get_task_list(taskId=taskId)[0]
    taskId = task_info["id"]
    goods_id = task_info["goods_id"]
    goodsIdList.append({
        "taskId": taskId,
        "goodsId": goods_id
    })
    sleep(0.1)
print(goodsIdList)
```
